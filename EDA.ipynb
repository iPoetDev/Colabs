### Biases Types and their Idenitificaton

#### Families of Biases


1. `Data and sampling biases`: These arise from how data is collected, sampled, or represented before and during EDA.
2. `Measurement and feature biases`: These come from how variables are defined, captured, or engineered before or during EDA.
3. `Cognitive and interpretive biases in EDA`: These are human biases that shape how analysts explore, clean, and interpret data
4. `Social and structural biases`: These reflect broader societal inequities and stereotypes encoded in the data and carried through EDA.
5. `Algorithmic and modeling biases (EDA-adjacent)`: These appear when EDA choices interact with model design and training.

> ---

#### Bias Types

| Bias type | Brief definition |
|---|---|
| Selection bias | The dataset is chosen in a way that does not represent the real population. |
| Sampling bias | Some groups are overrepresented or underrepresented in the sample. |
| Representation bias | Important user groups or cases are missing or too sparse in the data. |
| Historical bias | Past inequalities or outdated patterns are baked into the data. |
| Temporal bias | Data reflects one time period and does not hold across changing conditions. |
| Measurement bias | Features, instruments, or logging methods capture reality inaccurately or unevenly. |
| Labeling / annotation bias | Human labels reflect inconsistent judgment or subjective assumptions. |
| Reporting bias | Some events are more likely to be recorded, surfaced, or emphasized than others. |
| Exclusion bias | Relevant data is dropped during collection, cleaning, or preprocessing. |
| Missing-data bias | Values are missing in non-random ways that distort analysis. |
| Survivorship bias | Only the “successful” or remaining cases are analyzed, while failures disappear. |
| Participation / nonresponse bias | People who opt in or respond differ systematically from those who do not. |
| Coverage bias | The data source misses parts of the target population entirely. |
| Proxy bias | A substitute variable stands in for a concept and smuggles in hidden unfairness. |
| Omitted-variable bias | An important explanatory variable is left out, skewing conclusions. |
| Cognitive bias | Human judgment in EDA is influenced by mental shortcuts or assumptions. |
| Confirmation bias | Analysts notice and trust patterns that support what they already believe. |
| Automation bias | People trust automated tools or model outputs too readily, even when flawed. |
| Implicit bias | Unconscious assumptions shape how data is collected, labeled, or interpreted. |
| Observer bias | Annotators or reviewers influence outcomes through their own expectations. |
| Aggregation bias | Different groups are wrongly combined and treated as if one pattern fits all. |
| Evaluation bias | The test set or metrics give a misleading picture of model quality or fairness. |
| Algorithmic bias | The model or optimization method amplifies unfair patterns in the data. |
| Feedback-loop bias | Model outputs affect future data collection, reinforcing the same skew over time. |
| Social / structural bias | Broader societal inequities are reflected in the data and then reproduced by the model. |
| Stereotyping bias | The system learns and repeats oversimplified group-based assumptions. |
| Prejudice bias | The model reflects discriminatory human judgments embedded in the data. |

> ---

### Visibility of Biases

Biases especially visible in EDA of text, image, and behavioral data
These are common in modern multimodal AI datasets.

- `Reporting / selection bias in content`
    Text and media datasets often overrepresent certain topics or groups (e.g., English content, Western news), so EDA on word frequencies or embeddings reflects cultural skew.

- `Label distribution / class prior bias`
    Imbalanced labels (e.g., more positive than negative sentiment, more non‑toxic than toxic comments) can make baseline models look strong but hide poor performance on minority classes.

- `Context loss bias`
    EDA that strips or ignores context (e.g., sentence‑level analysis without conversation context) can mischaracterize groups or topics, biasing downstream classifiers.

### Bias Identification: Temporal (Time) Bias
AI models can be biased if they only see data from specific time periods. This plot will show which years have the most coverage, helping identify if the model will be biased toward recent trends.

````python
# Identify numeric year columns
year_cols = [col for col in df_goal3.columns if col.isdigit()]
data_by_year = df_goal3[year_cols].notnull().sum()

plt.figure(figsize=(12, 5))
data_by_year.plot(kind='line', marker='o', color='green', linewidth=2)
plt.title('Temporal Data Availability for Goal 3 (Health)')
plt.xlabel('Year')
plt.ylabel('Number of Non-Null Observations')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
````

### Bias Mitigation: Handling Missing Data (Imputation)

> Since we identified significant gaps (**Temporal Bias**), we can mitigate this by filling missing values.

- A common approach for *SDG time-series data* is `'Forward Fill'`,
    - where we assume the last reported value remains valid
    - for subsequent years until a new update is recorded.

````python
# Apply forward fill across the year columns (axis=1) for each row
df_goal3_mitigated = df_goal3.copy()
year_cols = [col for col in df_goal3.columns if col.isdigit()]
df_goal3_mitigated[year_cols] = df_goal3_mitigated[year_cols].ffill(axis=1)

# Check if availability improved for a sample year (e.g., 2017)
before = df_goal3['2017'].notnull().sum()
after = df_goal3_mitigated['2017'].notnull().sum()

print("Forward Fill Mitigation")
print(f"Observations in 2017 before mitigation: {before}")
print(f"Observations in 2017 after mitigation: {after}")
````

### Bias Impact Discussion Framework (Part 2b)

Following our EDA, we have identified two primary forms of bias in the SDG Goal 3 dataset. Below is a framework for discussing their impacts on AI models:

#### 1. Regional / Geographic Bias
*   **Observed Bias:** Certain countries or regions (e.g., Arab World, specific developed regions) have significantly more reported observations than others.
*   **Impact on AI:** Models trained on this data may exhibit higher accuracy for over-represented regions while failing to generalize to under-represented ones (often Global South nations). This leads to **algorithmic exclusion**, where health interventions or resource allocations suggested by the model are less effective for the populations most in need.

#### 2. Temporal / Time-Series Bias
*   **Observed Bias:** A significant drop in data availability for very recent years (e.g., 2017-2018) and sparse reporting in the early 1990s.
*   **Impact on AI:** Models may become **historically anchored**, relying on patterns from a decade ago that do not reflect modern medical advancements or shifting health crises (like emerging pandemics). Predictive models might lose accuracy in forecasting current health trends if the training window is dominated by older, more complete data blocks while lacking recent context.

#### 3. Ethical and Practical Implications
*   **Fairness:** Using biased data to train AI in healthcare can reinforce existing global health inequalities.
*   **Risk of Misdiagnosis:** If a diagnostic AI is trained on data from a region with specific genetic or environmental profiles, it may misdiagnose patients from regions with different health determinants.
*   **Mitigation Necessity:** The use of 'Forward Fill' (imputation) is a technical necessity to maintain longitudinal integrity, but we must acknowledge that it assumes stability where volatility might actually exist, potentially masking recent health regressions.
